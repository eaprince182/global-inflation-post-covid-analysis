```python
import pandas as pd

# Load dataset
df = pd.read_csv("global_inflation_post_covid.csv")

# Display first rows
print(df.head())

# Check dataset information
print(df.info())

# Check missing values
print(df.isnull().sum())

# Check duplicates
print("Duplicates", df.duplicated().sum())

# Clean column names
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

# Convert date column
df["date"] = pd.to_datetime(df["date"])

# Remove duplicates
df = df.drop_duplicates()

# Show cleaned dataset info
print(df.info())

# Save cleaned dataset
df.to_csv("cleaned_global_inflation.csv", index=False)

print("Cleaned Dataset Sucessfully.")

```
